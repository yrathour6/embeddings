```python
#pip install sentence-transformers
from sentence_transformers import SentenceTransformer
import numpy as np

# --- Step 1: Load a pretrained embedding model (runs locally, CPU is fine) ---
model = SentenceTransformer('all-MiniLM-L6-v2')

# --- Step 2: Corpus ---
docs = [
    "The cat sat on the mat",
    "Dogs are great companions",
    "I love programming in Python",
    "Python is a popular programming language",
    "Cats and dogs are common pets"
]

# --- Step 3: Generate real embeddings ---
embeddings = model.encode(docs)  # shape: (5, 384) — each doc is now a 384-dim vector

# --- Step 4: Cosine similarity (same as before) ---
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b) + 1e-8)

# --- Step 5: Search ---
query = "I enjoy coding"   # note: shares ZERO words with any doc
query_vec = model.encode([query])[0]

scores = [cosine_similarity(query_vec, doc_vec) for doc_vec in embeddings]
ranked = sorted(zip(docs, scores), key=lambda x: x[1], reverse=True)

print(f"Query: {query}\n")
for doc, score in ranked:
    print(f"{score:.3f}  {doc}")
```
